# Exploring the Number 2

This notebook practices binary numbers, bitwise operations, two-variable functions, and conditional comparisons.

## 1. Binary Representation of Numbers

Python represents integers in base 2 using bits. The `bin()` function shows the binary form, while integer arithmetic still works normally.

```python
values = [0, 1, 2, 3, 4, 8]

for value in values:
    print(f"{value:>2} -> {bin(value)}")

# Binary arithmetic: adding 2 and 3 gives 5.
print("2 + 3 =", 2 + 3)
print("Binary result:", bin(2 + 3))
```

## 2. Bitwise Operations with 2

The number `2` is binary `10`. Used as a bitmask, it lets us inspect or change the second bit of another integer.

```python
number = 5
mask = 2

print("number:", number, bin(number))
print("mask:", mask, bin(mask))
print("AND :", number & mask, bin(number & mask))
print("OR  :", number | mask, bin(number | mask))
print("XOR :", number ^ mask, bin(number ^ mask))
print("2 << 1:", 2 << 1)
print("2 >> 1:", 2 >> 1)
```

## 3. Two-Variable Functions

A function can accept two inputs. Here, `f(x, y) = x^2 + y^2` combines two squared values.

```python
def f(x, y):
    return x**2 + y**2

examples = [(0, 2), (1, 2), (2, 2), (3, 2)]
for x, y in examples:
    print(f"f({x}, {y}) = {f(x, y)}")
```

## 4. Comparing Values to 2

Conditional statements make decisions by comparing a value with a reference number.

```python
def compare_to_two(value):
    if value < 2:
        return "less than 2"
    if value == 2:
        return "equal to 2"
    return "greater than 2"

for value in [1, 2, 3]:
    print(f"{value} is {compare_to_two(value)}")
```
